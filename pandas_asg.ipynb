```python
"""1.Create a pandas DataFrame from the CSV and from Python lists/dicts."""

import pandas as pd 

#from csv 
df=pd.read_csv(r"D:\Downloads\sales_sample.csv")
print("Data from Csv file:\n")
print(df.head())

#from lists
data_lists = [
    [1, "2025-11-01", 5.25],    
    [2, "2025-11-02", 1.2],
    [3, "2025-11-03", 3.5],
]
df_lists = pd.DataFrame(data_lists, columns=["id", "Date", "total"])
print("\nData from Lists:\n")
print(df_lists)

#from dicts
data_dicts = [
    {"id": 1, "Date": "2025-11-01", "total": 5.25},
    {"id": 2, "Date": "2025-11-02", "total": 1.2},
    {"id": 3, "Date": "2025-11-03", "total": 3.5},  
]
df_dicts = pd.DataFrame(data_dicts)
print("\nData from Dicts:\n")
print(df_dicts)


```

```python
"""2.Filter and subset rows/columns (e.g., sales for a store, date ranges, high-value transactions)."""

# Filter sales for a specific store
store_id=3
filtered_sales=df[df['store_id']==3]
print("Sales for Store Id 3:\n")
print(filtered_sales.head())


# Filter sales within a date range
start_date="2025-10-01"
end_date="2025-10-15"

df['date']=pd.to_datetime(df['date'])

filtered_date=df[(df['date']>=start_date) & (df['date']<=end_date)]
print("\nData from date 2025-10-01 to 2025-10-15:\n")
print(filtered_date.head())

#high-value transactions
high_value=df[df['total']>30]
print("\nHigh value transactions(total above 30) :\n")
print(high_value.head())

#select specific columns
selected_col=df[['store_id','quantity','total']]
print("\nSelected Columns(store_id, quantity, total):\n")
print(selected_col.head())
```

```python
"""3.Compute descriptive statistics and group summaries (totals by store, average basket)."""

#descriptive statistics
print("Descriptive statistics:\n")
print(df.describe())

#group summaries
#totals by store
total_by_store=df.groupby('store_id')['total'].sum()
print("\nTotal sales by store:\n")
print(total_by_store.reset_index())

#average basket
avg_basket=df.groupby('store_id')['total'].mean()
print("\nAverage basket size:\n")
print(avg_basket.reset_index())
```

```python
"""4.Perform simple cleaning (handle missing values, fix data types, drop duplicates)."""

#check null values 
print(df.isnull().sum())

#handle missing values 
df['promo_code'].fillna('DISC10',inplace=True)

print(df.head())

#fix data types
df['date']=pd.to_datetime(df['date'])
df['store_id']=df['store_id'].astype(int)
df['total']=df['total'].astype(float)

#drop duplicates
df.drop_duplicates(inplace=True)

print("\nData after cleaning:\n")
print(df.info())
```

```python
#check if missing values are handled
print("Checking missing values after handling:\n")
print(df.isnull().sum())
```

```python
"""5.Build a mini ETL pipeline: read CSV → clean & transform → output JSON."""

def etl_pipeline(input_csv,output_json):

    #extract
    df=pd.read_csv(input_csv)

    #clean 
    print("Initial missing values:\n")
    print(df.isnull().sum())

    df.fillna({'promo_code':'DISC10'},inplace=True)

    df.drop_duplicates(inplace=True)

    print("\nData info after cleaning:\n")
    df.info()

    df['date']=pd.to_datetime(df['date'])
    df['store_id']=df['store_id'].astype(int)
    df['total']=df['total'].astype(float)

    #transform

    #add year column
    df['year']=df['date'].dt.year
    
    #store summary 
    summary = df.groupby('store_id')['total'].mean().reset_index()
    summary.rename(columns={'total': 'avg_sales_per_day'}, inplace=True)
    print(summary.head())

    df=df.merge(summary,on='store_id',how='left')

    #load
    df.to_json(output_json,orient='records',lines=True)
    print(f"\nData exported to {output_json}")

etl_pipeline(r"D:\Downloads\sales_sample.csv",r"D:\Documents\Chubb\Assignment1\sales_output.json")
    

     
```

```python
"""6.Complete a short hands-on test and an assignment to extend the pipeline."""

def etl_pipeline_extended(input_csv,output_json,top10_products_json,weekday_summary_json):
    #extract
    df=pd.read_csv(input_csv)

    #clean 
    print("Initial missing values:\n")
    print(df.isnull().sum())

    df.fillna({'promo_code':'DISC10'},inplace=True)

    df.drop_duplicates(inplace=True)

    print("\nData info after cleaning:\n")
    df.info()

    df['date']=pd.to_datetime(df['date'])
    df['store_id']=df['store_id'].astype(int)
    df['total']=df['total'].astype(float)

    #transform

    #add year column
    df['year']=df['date'].dt.year
    
    #store summary 
    summary = df.groupby('store_id')['total'].mean().reset_index()
    summary.rename(columns={'total': 'avg_sales_per_day'}, inplace=True)
    print(summary.head())

    df=df.merge(summary,on='store_id',how='left')

    #extension
    #1:top 10 products by sales
    top10_products=df.groupby('product_id')['total'].sum().reset_index()
    top10_products=top10_products.sort_values(by='total',ascending=False).head(10)

    print("\nTop 10 products by sales:\n")
    print(top10_products)
    top10_products.to_json(top10_products_json,orient='records',lines=True)
    print(f"\nTop 10 products exported to {top10_products_json}")

    #2:add weekday column
    df['weekday']=df['date'].dt.day_name()
    #weekday sales summary
    weekday_summary=df.groupby('weekday')['total'].sum().reset_index()
    print("\nWeekday sales summary:\n")
    print(weekday_summary)
    weekday_summary.to_json(weekday_summary_json,orient='records',lines=True)
    print(f"\nWeekday summary exported to {weekday_summary_json}")

    #3:calculate discount
    def calculate_discount(data):    
        if data['promo_code']=='DISC5':
            return data['total']*0.05
        elif data['promo_code']=='DISC10':
            return data['total']*0.10
        else:
            return 0.0  
    df['discount']=df.apply(calculate_discount,axis=1)
    print("\nData with discount column:\n")
    print(df.head())


    #load
    df.to_json(output_json,orient='records',lines=True)
    print(f"\nData exported to {output_json}")

etl_pipeline_extended(r"D:\Downloads\sales_sample.csv",
                      r"D:\Documents\Chubb\Assignment1\sales_output_extended.json",
                      r"D:\Documents\Chubb\Assignment1\top10_products.json",
                      r"D:\Documents\Chubb\Assignment1\weekday_summary.json")


```
